```python
# Univariate Plot for Gender Analysis
plt.figure(figsize=(7, 5))
sns.countplot(x='gender', data=df, palette='viridis')
plt.title('Distribution of Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
# Univariate plot for 'income'
plt.figure(figsize=(8, 5))
sns.histplot(df['income'], kde=True, bins=20)
plt.title('Distribution of Income')
plt.xlabel('Income')
plt.ylabel('Frequency')
plt.show()
```

```python
# Bivariate Plot Doctor Visit vs Age
plt.figure(figsize=(10, 6))
sns.scatterplot(x='age', y='visits', data=df, alpha=0.5)
plt.title('Doctor Visits vs. Age')
plt.xlabel('Age')
plt.ylabel('Number of Doctor Visits')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Bivariate Plot Doctor Visits vs Income
plt.figure(figsize=(10, 6))
sns.scatterplot(x='income', y='visits', data=df, alpha=0.5)
plt.title('Doctor Visits vs. Income')
plt.xlabel('Income')
plt.ylabel('Number of Doctor Visits')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Multivariate Plot
plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```

```python
gender_visits_mean = df.groupby('gender')['visits'].mean().reset_index()

plt.figure(figsize=(7, 5))
sns.barplot(x='gender', y='visits', data=gender_visits_mean, palette=['teal'])
plt.title('Mean Doctor Visits by Gender')
plt.xlabel('Gender')
plt.ylabel('Mean Number of Doctor Visits')
plt.show()
```

```python
illness_visits_mean = df.groupby('illness')['visits'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.lineplot(x='illness', y='visits', data=illness_visits_mean, marker='o')
plt.title('Mean Doctor Visits by Illness Level')
plt.xlabel('Illness Level')
plt.ylabel('Mean Number of Doctor Visits')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='gender', y='visits', data=df, palette='pastel', hue='gender', legend=False)
plt.title('Distribution of Doctor Visits by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Doctor Visits')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Plot 1: Univariate plot for 'age'
plt.figure(figsize=(8, 5))
sns.histplot(df['age'], kde=True, bins=20, palette='magma')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot 2: Mean Doctor Visits by Health Status
health_visits_mean = df.groupby('health')['visits'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='health', y='visits', data=health_visits_mean, palette='viridis', hue='health', legend=False)
plt.title('Mean Doctor Visits by Health Status')
plt.xlabel('Health Status')
plt.ylabel('Mean Number of Doctor Visits')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Plot 3: Mean Doctor Visits by Private Insurance Status
private_visits_mean = df.groupby('private')['visits'].mean().reset_index()

plt.figure(figsize=(7, 5))
sns.barplot(x='private', y='visits', data=private_visits_mean, palette='cividis', hue='private', legend=False)
plt.title('Mean Doctor Visits by Private Insurance Status')
plt.xlabel('Private Insurance (yes/no)')
plt.ylabel('Mean Number of Doctor Visits')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
